```python
import numpy as np
import matplotlib.pyplot as plt

cm = np.loadtxt("out/confusion_matrix.csv", delimiter=",").astype(int)
classes = [line.strip() for line in open("out/classes.txt", encoding="utf-8")]

plt.figure(figsize=(6,5))
plt.matshow(cm)
plt.title('Confusion matrix')
plt.colorbar()
plt.ylabel(' Giá trị thực')
plt.xlabel(' Giá trị dự đoán')
plt.show()

```
